I was going to do a calculation of whether or not the cloud would be gravitationally bound around the star if it were at its location, but I thought that might be unnecessary, but now that I reflect upon it again, it might actually kind of make sense to do... but no, not really, this just cements that the presence of a second one of these (SDSSJ1330+6435) that looks like identical, is the real damning evidence that this isn't the ISM because that seems like a bit of a leap to find two of these things.

```python
def get_real_sep(sep_arcsec, distance=distance):
    max_width=np.tan(sep_arcsec.to(u.radian))*distance
    return max_width.to(u.au)
```

```python
other_sep=get_real_sep(4.5*u.arcsec)
```

```python
print(other_sep)
```

```python
m_wd=0.56 *u.M_sun
```

```python
vesc= np.sqrt(2*const.G*m_wd/other_sep)
```

```python
print(vesc.to(u.km/u.s))
```

```python
potential_energy=-1*const.G*m_wd*0.01*u.M_sun/other_sep
```

```python
def kep3(m_total, sep):
    
    return np.sqrt((4*np.pi**2)/(const.G*(m_total))*(sep/2.)**3)
```

```python
j1458_period= kep3(0.56*u.M_sun*2, j1458_sep)
print(j1458_period.to(u.yr))
```
